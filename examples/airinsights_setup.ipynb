<h1>Getting Started with Air Insights</h1>

<h3> Generate a configuration file for your data </h3>
<p> Air Insights functions use a configuration file to identify column names, timestamp format, time zones, and which pollutants are present in the input data to ensure they are parsed correctly. The <code>build_config()</code> function can be used to create the configuration file and save it locally for future use.</p>

<details>
    <summary>Click here to expand a list detailing each configuration parameter.</summary>
    <br>
    <b>The following parameters are required:</b>
    <ul>
        <li>local_tz : Local timezone to convert the timestamp column to. For example: "America/Los_Angeles".</li>
            For more info, see <a href = https://en.wikipedia.org/wiki/List_of_tz_database_time_zones>Wikipedia's list of database time zones by clicking here</a>
        <li>timestamp_col : Name of the column containing date and time</li>
        <li>timestamp_format : Format of the timestamp column using python datetime syntax.
        <br>For example: "2026-01-01 09:27:11" -> "%Y-%m-%d %H:%M:%S". For more info, see <a href = https://docs.python.org/3/library/datetime.html#strftime-and-strptime-behavior>the python documentation by clicking here</a>
        <li>timestamp_tz : Timezone of the input timestamp column. Same formatting as local_tz. </li>
            For more info, see <a href = https://en.wikipedia.org/wiki/List_of_tz_database_time_zones>Wikipedia's list of database time zones by clicking here</a>
        <li>site_col : Name of the column containing unique identifiers for the monitoring sites</li>
        <li>lat_col : Name of the latitude column</li>
        <li>lon_col : Name of the longitude column</li>
        <li>pollutant_map : A dictionary specifying the pollutant(s) to analyze. The keys must be standard Air Insights pollutant names (e.g., "PM2.5", "NO2", "O3"), with values corresponding to each pollutant's name in the input data. If the input data is in WIDE format, the pollutants exist as column names. For LONG format data, they are found within a text column. For example:<br>
            <blockquote>
            <code>
            {<br>
            "PM2.5": "pm25_concentration", <br>
            "NO2": "no2_ppb", <br>
            "O3": "ozone <br>
            }
            </code>
            </blockquote>
        <li>value_col : When input data is in LONG format, the name of the column containing the measurement values.</li>
        <li>config_file : Provide a file path to save the configuration as a YAML file.</li>
</details>

```python
import airinsights as air

config_path = "../config/example_config_output.yaml" # path where config yaml will be saved
air.build_config(
    timestamp_col="Datetime",
    timestamp_tz="America/Los_Angeles",
    local_tz="America/Los_Angeles",
    site_col="Site",
    timestamp_format="%m/%d/%Y %H:%M",
    lat_col="Latitude",
    lon_col="Longitude",
    pollutant_map={
    "BC": "hourly_BC",
    },
    config_file=config_path
    ) 
```

<h3>Reading in data</h3>

<p>Now that your configuration file is created, we can read in and process data with the <code>read_aqdata_file()</code> function by supplying a path to the input data and the configuration file.</p>

We'll be using a sample dataset of 100 black carbon sensors deployed in West Oakland in 2017.


```python
example_data = "sample_data/oakland_2017_100x100_blackcarbon.csv"
input_data, config = air.read_aqdata_file(input_file=example_data,
                                          config_file=config_path)

# Preview data
input_data.head()
```
